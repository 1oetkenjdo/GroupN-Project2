### Define Constants

```python
cubic_yard_ft3 = 27  # 1 cubic yard is 27 cubic feet
unit_weight_water = 62.4  # Unit weight of water in lb/ft^3
```

### Create and Calculate Mix Designs

```python
# Create a reusable function to automate the mix design calculations
def mix_design(cement_wt, flyash_wt, silica_wt, scm_wt, wc_ratio, air_pct, fine_agg_pct, coarse_agg_pct, other_agg_pct, 
               cement_sg, flyash_sg, silica_sg, scm_sg, fine_agg_sg, coarse_agg_sg, other_agg_sg):

    # Calculate water weight based on total cementitious materials and w/c ratio
    water_wt = (cement_wt + flyash_wt + silica_wt + scm_wt) * wc_ratio
    
    # Calculate volumes (ft^3)
    cement_vol = cement_wt / (cement_sg * unit_weight_water)
    flyash_vol = flyash_wt / (flyash_sg * unit_weight_water)
    silica_vol = silica_wt / (silica_sg * unit_weight_water)
    scm_vol = scm_wt / (scm_sg * unit_weight_water)
    air_vol = (air_pct / 100) * cubic_yard_ft3
    water_vol = water_wt / unit_weight_water
    
    # Calculate total aggregate volume
    total_agg_vol = cubic_yard_ft3 - cement_vol - flyash_vol - silica_vol - scm_vol - air_vol - water_vol
    
    # Calculate aggregate weights
    fine_agg_wt = unit_weight_water * (fine_agg_pct / 100) * fine_agg_sg * total_agg_vol
    coarse_agg_wt = unit_weight_water * (coarse_agg_pct / 100) * coarse_agg_sg * total_agg_vol
    other_agg_wt = unit_weight_water * (other_agg_pct / 100) * other_agg_sg * total_agg_vol
    
    return (cement_wt, flyash_wt, silica_wt, scm_wt, fine_agg_wt, coarse_agg_wt, other_agg_wt, water_wt)
```

### User Input Mix Design
This code is the user input required mix design

```python
# User Input: Target Amounts
A = float(input("Enter the weight of cement per cubic yard: "))
B = float(input("Enter the weight of fly ash in cubic yard: "))
C = float(input("Enter the weight of silica fume per cubic yard: "))
D = float(input("Enter the weight of SCM per cubic yard: "))
E = float(input("Enter the target water/cement ratio: "))
F = float(input("Enter the target % of air content: "))
G = float(input("Enter the target % of fine aggregate: "))
H = float(input("Enter the target % of coarse aggregate: "))
I = float(input("Enter the target % of other aggregate: "))

# User Input: Specific Gravity Values
J = float(input("Enter the specific gravity of cement (MUST BE NON ZERO): "))
K = float(input("Enter the specific gravity of fly ash (MUST BE NON ZERO): "))
L = float(input("Enter the specific gravity of silica fume (MUST BE NON ZERO): "))
M = float(input("Enter the specific gravity of other SCM (MUST BE NON ZERO): "))
N = float(input("Enter the specific gravity of fine aggregate (MUST BE NON ZERO): "))
O = float(input("Enter the specific gravity of coarse aggregate (MUST BE NON ZERO): "))
P = float(input("Enter the specific gravity of other aggregate (MUST BE NON ZERO): "))

# Calculate mix design using the function
cement, fly_ash, silica_fume, other_scm, fine_aggregate, coarse_aggregate, other_aggregate, water = mix_design(
    A, B, C, D, E, F, G, H, I, J, K, L, M, N, O, P
)

# Calculate total mass for user input mix design
total_mass = cement + fly_ash + silica_fume + other_scm + fine_aggregate + coarse_aggregate + other_aggregate + water

# Print User Input Mix Design Results
print("\n-----------------------------------------")
print("  User Input Mix Design – Weight Summary")
print("      (1 Cubic Yard of Concrete)")
print("-----------------------------------------")
print("Cementitious Materials:")
print(f"  Cement:              {cement:8.2f} lb")
print(f"  Fly Ash:             {fly_ash:8.2f} lb")
print(f"  Silica Fume:         {silica_fume:8.2f} lb")
print(f"  Other SCM:           {other_scm:8.2f} lb")
print("-----------------------------------------")

print("Aggregates:")
print("-----------------------------------------")

print(f"  Fine Aggregate:      {fine_aggregate:8.2f} lb")
print(f"  Coarse Aggregate:    {coarse_aggregate:8.2f} lb")
print(f"  Other Aggregate:     {other_aggregate:8.2f} lb")

print("-----------------------------------------")

print(f"  Total Mass:            {total_mass:8.2f} lb/yd³")
print(f"  Water:                 {water:8.2f} lb")

print("-----------------------------------------")
```
